```python
import sys
import os
from pathlib import Path

CURRENT_DIR = os.getcwd()
FAXEN_ROOT = os.path.abspath(os.path.join(CURRENT_DIR, '..'))
sys.path.append(FAXEN_ROOT)
from packages import ROMCurve4BayesianInference

model = ROMCurve4BayesianInference(
    swell_root=Path(FAXEN_ROOT),
    train_data_rels=["datas/giesekus/uavg-0.1/train/curve4_y.txt"],
    model="giesekus",  # "oldroyd", "giesekus", or "ptt"
    mode="full_curve",  # or "full_curve"
    rom_method="gpr",
    scaler="standard",
    eps=1e-6,
    sigma_noise_percent=0.1,
    thin=2,
    sigma_bias=None,
    true_theta=[3.11, 0.87, 0.25])

model.load_data(filename="datas/ptt/uavg-0.1/infer/lambda_5.5_beta_0.25_alpha_0.25.txt", u_avg_obs=0.1)

model.plot_data()  
model.build_rom()
model.plot_rom_prediction()
model.plot_prior()

model.run_mcmc(nwalkers = 10, nsteps = 10000, warmup = True)
model.print_diagnostics()
model.plot_posterior_predictive()
model.plot_corner_all()
model.plot_trace_all()
```
